```python
# ============================================
# DETEKSI PROMOSI JUDI ONLINE
# Machine Learning - Naive Bayes + TF-IDF
# ============================================

# 1. Import library
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 2. Membuat dataset sederhana
data = {
    'teks': [
        'Daftar sekarang situs gacor bonus member baru',
        'Deposit 20 ribu dapat bonus besar',
        'Main slot hari ini jackpot jutaan rupiah',
        'Situs judi online terpercaya dan mudah menang',
        'Bonus deposit 100 persen khusus member baru',
        'Link slot gacor malam ini',
        'Daftar judi online dan dapatkan cashback',
        'Menang besar hanya dengan modal kecil',
        'Promo taruhan bola malam ini',
        'Jackpot besar tersedia di situs kami',

        'Besok ada kuliah jam 8 pagi',
        'Silakan kumpulkan tugas sebelum hari Jumat',
        'Rapat organisasi dilaksanakan sore ini',
        'Saya sedang belajar pemrograman Python',
        'Jadwal ujian sudah diumumkan oleh kampus',
        'Ayo makan siang bersama di kantin',
        'Dosen memberikan tugas kelompok hari ini',
        'Perpustakaan buka sampai jam lima sore',
        'Saya membeli laptop baru untuk kuliah',
        'Besok kita presentasi di kelas'
    ],

    'label': [
        'Judi Online',
        'Judi Online',
        'Judi Online',
        'Judi Online',
        'Judi Online',
        'Judi Online',
        'Judi Online',
        'Judi Online',
        'Judi Online',
        'Judi Online',

        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi',
        'Bukan Judi'
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

# 3. Memisahkan data training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    df['teks'],
    df['label'],
    test_size=0.3,
    random_state=42
)

# 4. Mengubah teks menjadi angka dengan TF-IDF
tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words=None
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# 5. Membuat dan melatih model Naive Bayes
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

# 6. Prediksi data testing
y_pred = model.predict(X_test_tfidf)

# 7. Evaluasi model
akurasi = accuracy_score(y_test, y_pred)

print("\n==============================")
print("HASIL EVALUASI MODEL")
print("==============================")
print("Akurasi:", akurasi)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# 8. Fungsi untuk mendeteksi teks baru
def deteksi_judol(teks):
    teks_tfidf = tfidf.transform([teks])
    hasil = model.predict(teks_tfidf)[0]
    probabilitas = model.predict_proba(teks_tfidf).max()

    print("\n==============================")
    print("HASIL DETEKSI")
    print("==============================")
    print("Teks :", teks)
    print("Hasil:", hasil)
    print("Tingkat keyakinan:", round(probabilitas * 100, 2), "%")

# 9. Contoh pengujian
deteksi_judol("daftar situs slot gacor bonus besar hari ini")

deteksi_judol("besok saya ada kuliah jam 9 pagi")
```
